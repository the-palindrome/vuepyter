```python
import numpy as np
from mlfz import Scalar

n_samples = 100
xs = np.random.rand(n_samples)
ys = 0.3 * xs + 0.8 + np.random.normal(size=n_samples, scale=0.05)
```

```python
await micropip.install("matplotlib")
```

```python
import matplotlib.pyplot as plt

with plt.style.context("seaborn-v0_8"):
    plt.figure(figsize=(5, 5))
    plt.scatter(xs, ys)
    plt.show()
```

```python
from mlfz.nn import Model


class Linear(Model):
    def __init__(self):
        self.a = Scalar(1)
        self.b = Scalar(1)

    def forward(self, x):
        return self.a * x + self.b
    
    def parameters(self):
        return {"a": self.a, "b": self.b}
```

```python
linear = Linear()
```

```python
from mlfz.nn.scalar.loss import mean_squared_error


n_steps = 100
lr = 0.2


for i in range(1, n_steps + 1):
    preds = [linear(x) for x in xs]
    l = mean_squared_error(preds, ys)
    l.backward()
    linear.gradient_update(lr=lr)
    
    if i == 1 or i % 10 == 0:
        print(f"step no. {i}, loss = {l.value}\n")
```
