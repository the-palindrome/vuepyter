# Game of Life + Vuepyter

```python
class GameOfLife:
    """Conway's Game of Life with helpers for notable classic patterns."""

    _PATTERN_RLES = {
        # Still lifes
        "block": "2o$2o!",
        "beehive": "b2ob$o2bo$b2o!",
        "loaf": "b2ob$o2bo$bobo$2bo!",
        "boat": "2ob$obo$bo!",
        "tub": "bob$obo$bo!",
        # Oscillators
        "blinker": "3o!",
        "toad": "b3o$3o!",
        "beacon": "2o2b$o3b$3bo$2b2o!",
        "pulsar": "2b3o3b3o2b2$o4bobo4bo$o4bobo4bo$o4bobo4bo$2b3o3b3o2b2$2b3o3b3o2b$o4bobo4bo$o4bobo4bo$o4bobo4bo2$2b3o3b3o!",
        "pentadecathlon": "2bo4bo2b$2ob4ob2o$2bo4bo!",
        # Spaceships
        "glider": "bob$2bo$3o!",
        "lwss": "bo2bo$o4b$o3bo$4o!",
        "mwss": "3bo2b$bo3bo$o5b$o4bo$5o!",
        "hwss": "3b2o2b$bo4bo$o6b$o5bo$6o!",
        # Methuselahs
        "r_pentomino": "b2o$2ob$bo!",
        "acorn": "bo5b$3bo3b$2o2b3o!",
        "diehard": "6bob$2o6b$bo3b3o!",
        # Guns
        "gosper_glider_gun": "24bo11b$22bobo11b$12b2o6b2o12b2o$11bo3bo4b2o12b2o$2o8bo5bo3b2o14b$2o8bo3bob2o4bobo11b$10bo5bo7bo11b$11bo3bo20b$12b2o!",
    }

    def __init__(self, rows: int, cols: int):
        self.rows = rows
        self.cols = cols
        self.reset()

    def reset(self):
        """Reset the board to all white squares (dead cells)."""
        self.grid = [[0 for _ in range(self.cols)] for _ in range(self.rows)]

    def set_cell(self, row: int, col: int, color: str):
        """Set one square to 'black' (alive) or 'white' (dead)."""
        if not (0 <= row < self.rows and 0 <= col < self.cols):
            raise IndexError("Cell is out of bounds")
        if color not in ("black", "white"):
            raise ValueError("color must be 'black' or 'white'")

        self.grid[row][col] = 1 if color == "black" else 0

    def _place_rle(self, row: int, col: int, rle: str):
        """Place an RLE pattern with (row, col) as the top-left anchor."""
        r = row
        c = col
        run = 0

        for ch in rle:
            if ch.isdigit():
                run = run * 10 + int(ch)
                continue

            count = run if run else 1
            run = 0

            if ch == "b":
                c += count
            elif ch == "o":
                for _ in range(count):
                    self.set_cell(r, c, "black")
                    c += 1
            elif ch == "$":
                r += count
                c = col
            elif ch == "!":
                break

    def add_pattern(self, name: str, row: int, col: int):
        """Add a named notable pattern at the given anchor coordinates."""
        key = name.strip().lower().replace("-", "_").replace(" ", "_")
        if key not in self._PATTERN_RLES:
            raise ValueError(f"Unknown pattern: {name}")
        self._place_rle(row, col, self._PATTERN_RLES[key])

    # Still lifes
    def set_block(self, row: int, col: int):
        self.add_pattern("block", row, col)

    def set_beehive(self, row: int, col: int):
        self.add_pattern("beehive", row, col)

    def set_loaf(self, row: int, col: int):
        self.add_pattern("loaf", row, col)

    def set_boat(self, row: int, col: int):
        self.add_pattern("boat", row, col)

    def set_tub(self, row: int, col: int):
        self.add_pattern("tub", row, col)

    # Oscillators
    def set_blinker(self, row: int, col: int):
        self.add_pattern("blinker", row, col)

    def set_toad(self, row: int, col: int):
        self.add_pattern("toad", row, col)

    def set_beacon(self, row: int, col: int):
        self.add_pattern("beacon", row, col)

    def set_pulsar(self, row: int, col: int):
        self.add_pattern("pulsar", row, col)

    def set_pentadecathlon(self, row: int, col: int):
        self.add_pattern("pentadecathlon", row, col)

    # Spaceships
    def set_glider(self, row: int, col: int):
        self.add_pattern("glider", row, col)

    def set_lwss(self, row: int, col: int):
        self.add_pattern("lwss", row, col)

    def set_mwss(self, row: int, col: int):
        self.add_pattern("mwss", row, col)

    def set_hwss(self, row: int, col: int):
        self.add_pattern("hwss", row, col)

    # Methuselahs
    def set_r_pentomino(self, row: int, col: int):
        self.add_pattern("r_pentomino", row, col)

    def set_acorn(self, row: int, col: int):
        self.add_pattern("acorn", row, col)

    def set_diehard(self, row: int, col: int):
        self.add_pattern("diehard", row, col)

    # Guns
    def set_gosper_glider_gun(self, row: int, col: int):
        self.add_pattern("gosper_glider_gun", row, col)

    def step(self):
        """Advance the simulation by one step using Conway's rules."""
        next_grid = [[0 for _ in range(self.cols)] for _ in range(self.rows)]

        for row in range(self.rows):
            for col in range(self.cols):
                live_neighbors = 0
                for dr in (-1, 0, 1):
                    for dc in (-1, 0, 1):
                        if dr == 0 and dc == 0:
                            continue
                        nr, nc = row + dr, col + dc
                        if 0 <= nr < self.rows and 0 <= nc < self.cols:
                            live_neighbors += self.grid[nr][nc]

                is_alive = self.grid[row][col] == 1
                if is_alive and live_neighbors in (2, 3):
                    next_grid[row][col] = 1
                elif not is_alive and live_neighbors == 3:
                    next_grid[row][col] = 1

        self.grid = next_grid

    def __str__(self):
        # Use ASCII characters so board state is easy to read in plain output.
        return "\n".join(
            "".join("#" if cell else "." for cell in row)
            for row in self.grid
        )
```

The `GameOfLife` object now includes helper methods for notable classic components (still lifes, oscillators, spaceships, methuselahs, and Gosper's glider gun).

Let's initialize a 128 x 128 board with a centered Gosper glider gun, a famous pattern that produces rich geometric structure over time:

```python
game = GameOfLife(rows=128, cols=128)

# Famous pattern: Gosper's glider gun (period-30 emitter of gliders).
# The anchor is chosen so the gun sits near the board center.
game.set_gosper_glider_gun(59, 46)
```

Execute this cell below to advance the state by a single step:

```python
game.step()   
```
