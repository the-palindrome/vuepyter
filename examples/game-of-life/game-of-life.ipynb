# Minimal Game of Life Example

`GameOfLife` controls the simulation with `reset()`, `step()`, and `set_cell()` to manually set a square to black or white.

```python
class GameOfLife:
    def __init__(self, rows: int, cols: int):
        self.rows = rows
        self.cols = cols
        self.reset()

    def reset(self):
        """Reset the board to all white squares (dead cells)."""
        self.grid = [[0 for _ in range(self.cols)] for _ in range(self.rows)]

    def set_cell(self, row: int, col: int, color: str):
        """Set one square to 'black' (alive) or 'white' (dead)."""
        if not (0 <= row < self.rows and 0 <= col < self.cols):
            raise IndexError("Cell is out of bounds")
        if color not in ("black", "white"):
            raise ValueError("color must be 'black' or 'white'")

        self.grid[row][col] = 1 if color == "black" else 0

    def step(self):
        """Advance the simulation by one step using Conway's rules."""
        next_grid = [[0 for _ in range(self.cols)] for _ in range(self.rows)]

        for row in range(self.rows):
            for col in range(self.cols):
                live_neighbors = 0
                for dr in (-1, 0, 1):
                    for dc in (-1, 0, 1):
                        if dr == 0 and dc == 0:
                            continue
                        nr, nc = row + dr, col + dc
                        if 0 <= nr < self.rows and 0 <= nc < self.cols:
                            live_neighbors += self.grid[nr][nc]

                is_alive = self.grid[row][col] == 1
                if is_alive and live_neighbors in (2, 3):
                    next_grid[row][col] = 1
                elif not is_alive and live_neighbors == 3:
                    next_grid[row][col] = 1

        self.grid = next_grid

    def __str__(self):
        # Use ASCII characters so board state is easy to read in plain output.
        return "\n".join(
            "".join("#" if cell else "." for cell in row)
            for row in self.grid
        )
```

```python
# Make a small glider pattern by setting individual squares.
game = GameOfLife(rows=256, cols=256)
game.set_cell(128, 129, "black")
game.set_cell(129, 130, "black")
game.set_cell(130, 128, "black")
game.set_cell(130, 129, "black")
game.set_cell(130, 130, "black")
```
